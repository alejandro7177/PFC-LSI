```python
import pandas as pd

df_general = pd.read_csv("../data/2_processed/questions_generals_reranked.csv")

df_specifics = pd.read_csv("../data/2_processed/questions_specific_reranked.csv")
df_general['specific'] = 0
df_specifics['specific'] = 1
```

```python
df_specifics.__len__()
```

```python
df_general.__len__()
```

```python
df_resultado = pd.concat([df_general, df_specifics.sample(n=156)], ignore_index=True)
df_resultado
```

```python
cols = ['doc_id_1','doc_id_2', 'doc_id_3', 'doc_id_4', 'doc_id_5', 'doc_id']
df_resultado['ids_docs'] = df_resultado[cols].apply(lambda x: x.dropna().tolist(), axis=1)
df_resultado
```

```python
df_resultado[['ids_docs', 'question']]

df_resultado.to_csv("../data/2_processed/questions_finals.csv")
df_resultado[['ids_docs', 'question', 'specific']].to_csv('../data/2_processed/question_finals(res).csv')
```
