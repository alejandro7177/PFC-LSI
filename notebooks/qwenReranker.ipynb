```python
from sentence_transformers import CrossEncoder

model = CrossEncoder("Qwen/Qwen3-Reranker-4B")
```

```python
query = "What is the capital of China?"
documents = [
    "The capital of China is Beijing.",
    "Gravity is a force that attracts two bodies towards each other. It gives weight to physical objects and is responsible for the movement of planets around the sun.",
]

pairs = [(query, doc) for doc in documents]
scores = model.predict(pairs)
print(scores)
# [  6.4375 -14.375 ]

rankings = model.rank(query, documents)
print(rankings)
```

```python
import pandas as pd

df = pd.read_csv("../data/3_gold/questions_v2_rrf_union.csv")
```

```python
N = 10
for index, row in df.sample(n=N).iterrows():
    print(f"**** {index} ****")
    query = row['question']
    docs  = [
        row[f'abstract_{i}'] for i in range(1,6) 
    ]
    scores = model.rank(query, docs)
    for score in scores:
        print(f"Index: {score['corpus_id']}")
        print(f"Score: {score['score']:.4f}")
        print(f"Document: {docs[score['corpus_id']][:100]}...")
        print("-" * 100)
    print("*"*120
    )
    input("Enter para continuar")
```

```python
if "max_score" not in df.columns:
    df["max_score"] = None
if "min_score" not in df.columns:
    df["min_score"] = None
```

```python
import numpy as np

def tang_hiperbolica(val:float)->float:
    return np.tanh(0.2 * val)
```

```python
for index, row in df.iterrows():
    print(f"**** {index} ****")
    query = row["question"]
    docs = [row[f"abstract_{i}"] for i in range(1, 6)]

    scores = model.rank(query, docs)

    # Extraer los puntajes
    lista_scores = [tang_hiperbolica(s["score"]) for s in scores]

    # Guardar los valores directamente en la fila del DataFrame usando .at
    df.at[index, "max_score"] = max(lista_scores)
    df.at[index, "min_score"] = min(lista_scores)
```

```python
df['max_score'] = df['max_score'].astype(float)
df['min_score'] = df['min_score'].astype(float)

df.describe()
```

```python
df_specific_questions = pd.read_csv('../data/3_gold/eval_retrieval_qwen_kw.csv')
df_dataset = pd.read_csv('../data/1_raw/dataset.csv')
```

```python
resultados = []
for index, row in df_specific_questions.sample(n=500).iterrows():
    query = row['question']
    row_doc = df_dataset.loc[df_dataset['doc_id'] == row['doc_id']]
    abstract = row_doc['abstract'].values[0]

    scores = model.rank(query, [abstract])
    resultados.append(
        {
            "doc_id": row['doc_id'],
            "question": query,
            "abstract": abstract,
            "score": tang_hiperbolica(scores[0]['score']),
        }
    )

df_resultados = pd.DataFrame(resultados)
df_resultados
```

```python
df_resultados_f = df_resultados[df_resultados['score']>0.50]
df_resultados_f.describe()
```
